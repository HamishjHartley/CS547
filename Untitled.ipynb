```python
# Step 2: Fitness Function
# 2.1 Fault coverage

def fault_coverage(selected_tests):
    faults_covered = set()
    for faults, _ in selected_tests:
        faults_covered.update(idx for idx, fault in enumerate(faults) if fault == 1)
    return len(faults_covered)

fault_coverage(matrix)
```

```python
# 2.2 Total execution time
def total_execution_time(selected_tests):
    return sum(time for _, time in selected_tests)

def test_set_size(selected_tests):
    return len(selected_tests)
```
